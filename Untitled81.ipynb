```python
import pandas as pd
import numpy as np

# Load cleaned dataset

file_name = "Cleaned_Sample_Dataset.xlsx"

df = pd.read_excel(file_name)

df["Event_Date"] = pd.to_datetime(df["Event_Date"], errors="coerce")


# Step 1: Basic Data Overview

total_rows = len(df)
total_columns = len(df.columns)
total_users = df["User_ID"].nunique()

data_overview = pd.DataFrame({
    "Metric": [
        "Total Rows",
        "Total Columns",
        "Unique Users"
    ],
    "Value": [
        total_rows,
        total_columns,
        total_users
    ]
})


# Step 2: Create Performance Metrics

df["Conversion_Rate"] = np.where(
    df["Requests"] > 0,
    (df["Successful_Requests"] / df["Requests"]) * 100,
    0
)

df["Abandonment_Rate"] = np.where(
    df["Requests"] > 0,
    (df["Failed_Requests"] / df["Requests"]) * 100,
    0
)

df["Error_Rate"] = np.where(
    df["Requests"] > 0,
    (df["Failed_Requests"] / df["Requests"]) * 100,
    0
)

df["Revenue_Impact"] = (
    df["Failed_Requests"] *
    df["Revenue"] /
    df["Successful_Requests"].replace(0, np.nan)
)

df["Revenue_Impact"] = df["Revenue_Impact"].fillna(0)

df["Profit"] = df["Revenue"] - df["Cost"]

df["Profit_Margin"] = np.where(
    df["Revenue"] > 0,
    (df["Profit"] / df["Revenue"]) * 100,
    0
)

df["Revenue_Per_Success"] = np.where(
    df["Successful_Requests"] > 0,
    df["Revenue"] / df["Successful_Requests"],
    0
)

df["Cost_Per_Success"] = np.where(
    df["Successful_Requests"] > 0,
    df["Cost"] / df["Successful_Requests"],
    0
)

df["Profit_Per_Success"] = np.where(
    df["Successful_Requests"] > 0,
    df["Profit"] / df["Successful_Requests"],
    0
)

df["Month"] = df["Event_Date"].dt.to_period("M").astype(str)


# Step 3: Overall Performance KPIs

total_requests = df["Requests"].sum()
successful_requests = df["Successful_Requests"].sum()
failed_requests = df["Failed_Requests"].sum()

total_revenue = df["Revenue"].sum()
total_cost = df["Cost"].sum()
total_profit = df["Profit"].sum()

conversion_rate = (
    successful_requests / total_requests * 100
    if total_requests > 0 else 0
)

abandonment_rate = (
    failed_requests / total_requests * 100
    if total_requests > 0 else 0
)

error_rate = (
    failed_requests / total_requests * 100
    if total_requests > 0 else 0
)

average_latency = df["Latency_ms"].mean()
average_quality = df["Quality_Score"].mean()
average_recovery = df["Recovery_Time_min"].mean()

profit_margin = (
    total_profit / total_revenue * 100
    if total_revenue > 0 else 0
)

estimated_revenue_impact = df["Revenue_Impact"].sum()


overall_kpis = pd.DataFrame({
    "Metric": [
        "Total Users",
        "Total Requests",
        "Successful Requests",
        "Failed Requests",
        "Conversion Rate",
        "Abandonment Rate",
        "Error Rate",
        "Total Revenue",
        "Total Cost",
        "Total Profit",
        "Profit Margin",
        "Average Latency (ms)",
        "Average Quality Score",
        "Average Recovery Time (min)",
        "Estimated Revenue Impact"
    ],
    "Value": [
        total_users,
        total_requests,
        successful_requests,
        failed_requests,
        conversion_rate,
        abandonment_rate,
        error_rate,
        total_revenue,
        total_cost,
        total_profit,
        profit_margin,
        average_latency,
        average_quality,
        average_recovery,
        estimated_revenue_impact
    ]
})


# Step 4: Monthly Performance Analysis

monthly_analysis = df.groupby("Month").agg(
    Requests=("Requests", "sum"),
    Successful_Requests=("Successful_Requests", "sum"),
    Failed_Requests=("Failed_Requests", "sum"),
    Revenue=("Revenue", "sum"),
    Cost=("Cost", "sum"),
    Profit=("Profit", "sum"),
    Revenue_Impact=("Revenue_Impact", "sum"),
    Average_Latency_ms=("Latency_ms", "mean"),
    Average_Quality_Score=("Quality_Score", "mean")
).reset_index()

monthly_analysis["Conversion_Rate"] = (
    monthly_analysis["Successful_Requests"]
    / monthly_analysis["Requests"] * 100
)

monthly_analysis["Abandonment_Rate"] = (
    monthly_analysis["Failed_Requests"]
    / monthly_analysis["Requests"] * 100
)

monthly_analysis["Error_Rate"] = (
    monthly_analysis["Failed_Requests"]
    / monthly_analysis["Requests"] * 100
)

monthly_analysis["Profit_Margin"] = (
    monthly_analysis["Profit"]
    / monthly_analysis["Revenue"] * 100
)


# Step 5: Region Performance

region_analysis = df.groupby("Region").agg(
    Requests=("Requests", "sum"),
    Successful_Requests=("Successful_Requests", "sum"),
    Failed_Requests=("Failed_Requests", "sum"),
    Revenue=("Revenue", "sum"),
    Cost=("Cost", "sum"),
    Profit=("Profit", "sum"),
    Revenue_Impact=("Revenue_Impact", "sum"),
    Average_Latency_ms=("Latency_ms", "mean"),
    Average_Recovery_Time_min=("Recovery_Time_min", "mean"),
    Average_Quality_Score=("Quality_Score", "mean")
).reset_index()

region_analysis["Conversion_Rate"] = (
    region_analysis["Successful_Requests"]
    / region_analysis["Requests"] * 100
)

region_analysis["Abandonment_Rate"] = (
    region_analysis["Failed_Requests"]
    / region_analysis["Requests"] * 100
)

region_analysis["Error_Rate"] = (
    region_analysis["Failed_Requests"]
    / region_analysis["Requests"] * 100
)

region_analysis["Profit_Margin"] = (
    region_analysis["Profit"]
    / region_analysis["Revenue"] * 100
)


# Step 6: Platform Performance

platform_analysis = df.groupby("Platform").agg(
    Requests=("Requests", "sum"),
    Successful_Requests=("Successful_Requests", "sum"),
    Failed_Requests=("Failed_Requests", "sum"),
    Revenue=("Revenue", "sum"),
    Cost=("Cost", "sum"),
    Profit=("Profit", "sum"),
    Revenue_Impact=("Revenue_Impact", "sum"),
    Average_Latency_ms=("Latency_ms", "mean"),
    Average_Recovery_Time_min=("Recovery_Time_min", "mean"),
    Average_Quality_Score=("Quality_Score", "mean")
).reset_index()

platform_analysis["Conversion_Rate"] = (
    platform_analysis["Successful_Requests"]
    / platform_analysis["Requests"] * 100
)

platform_analysis["Abandonment_Rate"] = (
    platform_analysis["Failed_Requests"]
    / platform_analysis["Requests"] * 100
)

platform_analysis["Error_Rate"] = (
    platform_analysis["Failed_Requests"]
    / platform_analysis["Requests"] * 100
)

platform_analysis["Profit_Margin"] = (
    platform_analysis["Profit"]
    / platform_analysis["Revenue"] * 100
)


# Step 7: Version Performance

version_analysis = df.groupby("Version").agg(
    Requests=("Requests", "sum"),
    Successful_Requests=("Successful_Requests", "sum"),
    Failed_Requests=("Failed_Requests", "sum"),
    Revenue=("Revenue", "sum"),
    Cost=("Cost", "sum"),
    Profit=("Profit", "sum"),
    Revenue_Impact=("Revenue_Impact", "sum"),
    Average_Latency_ms=("Latency_ms", "mean"),
    Average_Quality_Score=("Quality_Score", "mean")
).reset_index()

version_analysis["Conversion_Rate"] = (
    version_analysis["Successful_Requests"]
    / version_analysis["Requests"] * 100
)

version_analysis["Abandonment_Rate"] = (
    version_analysis["Failed_Requests"]
    / version_analysis["Requests"] * 100
)

version_analysis["Error_Rate"] = (
    version_analysis["Failed_Requests"]
    / version_analysis["Requests"] * 100
)

version_analysis["Profit_Margin"] = (
    version_analysis["Profit"]
    / version_analysis["Revenue"] * 100
)


# Step 8: Latency Analysis

latency_analysis = df.groupby("Platform").agg(
    Average_Latency_ms=("Latency_ms", "mean"),
    Requests=("Requests", "sum"),
    Successful_Requests=("Successful_Requests", "sum"),
    Failed_Requests=("Failed_Requests", "sum"),
    Revenue=("Revenue", "sum"),
    Revenue_Impact=("Revenue_Impact", "sum")
).reset_index()

latency_analysis["Conversion_Rate"] = (
    latency_analysis["Successful_Requests"]
    / latency_analysis["Requests"] * 100
)

latency_analysis["Error_Rate"] = (
    latency_analysis["Failed_Requests"]
    / latency_analysis["Requests"] * 100
)


# Step 9: Latency Buckets

df["Latency_Band"] = pd.cut(
    df["Latency_ms"],
    bins=[0, 300, 500, 700, float("inf")],
    labels=[
        "Low Latency",
        "Medium Latency",
        "High Latency",
        "Very High Latency"
    ],
    include_lowest=True
)

latency_bands = df.groupby(
    "Latency_Band",
    observed=False
).agg(
    Requests=("Requests", "sum"),
    Successful_Requests=("Successful_Requests", "sum"),
    Failed_Requests=("Failed_Requests", "sum"),
    Revenue=("Revenue", "sum"),
    Revenue_Impact=("Revenue_Impact", "sum"),
    Average_Latency_ms=("Latency_ms", "mean")
).reset_index()

latency_bands["Conversion_Rate"] = (
    latency_bands["Successful_Requests"]
    / latency_bands["Requests"] * 100
)

latency_bands["Abandonment_Rate"] = (
    latency_bands["Failed_Requests"]
    / latency_bands["Requests"] * 100
)

latency_bands["Error_Rate"] = (
    latency_bands["Failed_Requests"]
    / latency_bands["Requests"] * 100
)


# Step 10: Error Analysis

error_analysis = df.groupby("Platform").agg(
    Requests=("Requests", "sum"),
    Failed_Requests=("Failed_Requests", "sum"),
    Successful_Requests=("Successful_Requests", "sum"),
    Revenue=("Revenue", "sum"),
    Revenue_Impact=("Revenue_Impact", "sum"),
    Average_Latency_ms=("Latency_ms", "mean")
).reset_index()

error_analysis["Error_Rate"] = (
    error_analysis["Failed_Requests"]
    / error_analysis["Requests"] * 100
)

error_analysis["Conversion_Rate"] = (
    error_analysis["Successful_Requests"]
    / error_analysis["Requests"] * 100
)


# Step 11: Revenue Impact Analysis

revenue_impact = df.groupby(
    ["Region", "Platform"]
).agg(
    Requests=("Requests", "sum"),
    Successful_Requests=("Successful_Requests", "sum"),
    Failed_Requests=("Failed_Requests", "sum"),
    Revenue=("Revenue", "sum"),
    Revenue_Impact=("Revenue_Impact", "sum"),
    Average_Latency_ms=("Latency_ms", "mean")
).reset_index()

revenue_impact["Error_Rate"] = (
    revenue_impact["Failed_Requests"]
    / revenue_impact["Requests"] * 100
)

revenue_impact["Conversion_Rate"] = (
    revenue_impact["Successful_Requests"]
    / revenue_impact["Requests"] * 100
)


# Step 12: Bottleneck Analysis

bottleneck_analysis = df.groupby(
    ["Region", "Platform"]
).agg(
    Requests=("Requests", "sum"),
    Average_Latency_ms=("Latency_ms", "mean"),
    Error_Rate=("Error_Rate", "mean"),
    Conversion_Rate=("Conversion_Rate", "mean"),
    Abandonment_Rate=("Abandonment_Rate", "mean"),
    Revenue_Impact=("Revenue_Impact", "sum"),
    Average_Quality_Score=("Quality_Score", "mean")
).reset_index()


# Step 13: Performance Priority Score

bottleneck_analysis["Latency_Score"] = (
    bottleneck_analysis["Average_Latency_ms"]
    / bottleneck_analysis["Average_Latency_ms"].max()
    * 100
)

bottleneck_analysis["Error_Score"] = (
    bottleneck_analysis["Error_Rate"]
    / bottleneck_analysis["Error_Rate"].max()
    * 100
)

bottleneck_analysis["Revenue_Impact_Score"] = (
    bottleneck_analysis["Revenue_Impact"]
    / bottleneck_analysis["Revenue_Impact"].max()
    * 100
)

bottleneck_analysis["Priority_Score"] = (
    bottleneck_analysis["Latency_Score"] * 0.35
    + bottleneck_analysis["Error_Score"] * 0.30
    + bottleneck_analysis["Revenue_Impact_Score"] * 0.35
)

bottleneck_analysis["Priority_Rank"] = (
    bottleneck_analysis["Priority_Score"]
    .rank(method="dense", ascending=False)
    .astype(int)
)

bottleneck_analysis = bottleneck_analysis.sort_values(
    "Priority_Rank"
)


# Step 14: Quality Analysis

quality_analysis = df.groupby("Platform").agg(
    Average_Quality_Score=("Quality_Score", "mean"),
    Average_Latency_ms=("Latency_ms", "mean"),
    Average_Recovery_Time_min=("Recovery_Time_min", "mean"),
    Requests=("Requests", "sum"),
    Failed_Requests=("Failed_Requests", "sum")
).reset_index()

quality_analysis["Error_Rate"] = (
    quality_analysis["Failed_Requests"]
    / quality_analysis["Requests"] * 100
)


# Step 15: Recovery Analysis

recovery_analysis = df.groupby("Region").agg(
    Average_Recovery_Time_min=("Recovery_Time_min", "mean"),
    Average_Latency_ms=("Latency_ms", "mean"),
    Error_Rate=("Error_Rate", "mean"),
    Requests=("Requests", "sum"),
    Revenue_Impact=("Revenue_Impact", "sum")
).reset_index()


# Step 16: Launch Status Analysis

launch_status_analysis = df.groupby("Launch_Status").agg(
    Requests=("Requests", "sum"),
    Successful_Requests=("Successful_Requests", "sum"),
    Failed_Requests=("Failed_Requests", "sum"),
    Revenue=("Revenue", "sum"),
    Profit=("Profit", "sum"),
    Revenue_Impact=("Revenue_Impact", "sum"),
    Average_Latency_ms=("Latency_ms", "mean"),
    Average_Quality_Score=("Quality_Score", "mean")
).reset_index()

launch_status_analysis["Conversion_Rate"] = (
    launch_status_analysis["Successful_Requests"]
    / launch_status_analysis["Requests"] * 100
)

launch_status_analysis["Error_Rate"] = (
    launch_status_analysis["Failed_Requests"]
    / launch_status_analysis["Requests"] * 100
)


# Step 17: Unit Economics

unit_economics = pd.DataFrame({
    "Metric": [
        "Revenue Per Successful Request",
        "Cost Per Successful Request",
        "Profit Per Successful Request",
        "Overall Profit Margin",
        "Estimated Revenue Impact"
    ],
    "Value": [
        total_revenue / successful_requests
        if successful_requests > 0 else 0,

        total_cost / successful_requests
        if successful_requests > 0 else 0,

        total_profit / successful_requests
        if successful_requests > 0 else 0,

        profit_margin,

        estimated_revenue_impact
    ]
})


# Step 18: Correlation Analysis

correlation_columns = [
    "Latency_ms",
    "Failed_Requests",
    "Successful_Requests",
    "Revenue",
    "Revenue_Impact",
    "Quality_Score",
    "Recovery_Time_min"
]

correlation_analysis = df[correlation_columns].corr()


# Step 19: High-Risk Areas

high_risk_areas = bottleneck_analysis[
    (
        bottleneck_analysis["Average_Latency_ms"]
        >= bottleneck_analysis["Average_Latency_ms"].median()
    )
    &
    (
        bottleneck_analysis["Error_Rate"]
        >= bottleneck_analysis["Error_Rate"].median()
    )
].copy()

high_risk_areas = high_risk_areas.sort_values(
    "Revenue_Impact",
    ascending=False
)


# Step 20: Top Revenue Impact Areas

top_revenue_impact = revenue_impact.sort_values(
    "Revenue_Impact",
    ascending=False
).head(10)


# Step 21: Business Insights

highest_latency_platform = platform_analysis.loc[
    platform_analysis["Average_Latency_ms"].idxmax(),
    "Platform"
]

highest_error_platform = platform_analysis.loc[
    platform_analysis["Error_Rate"].idxmax(),
    "Platform"
]

highest_revenue_impact_region = region_analysis.loc[
    region_analysis["Revenue_Impact"].idxmax(),
    "Region"
]

highest_priority_region = bottleneck_analysis.iloc[0]["Region"]

highest_priority_platform = bottleneck_analysis.iloc[0]["Platform"]

highest_latency_band = latency_bands.loc[
    latency_bands["Average_Latency_ms"].idxmax(),
    "Latency_Band"
]

insights = pd.DataFrame({
    "Business_Insight": [
        f"The overall conversion rate is {conversion_rate:.2f}%.",
        f"The overall error rate is {error_rate:.2f}%.",
        f"The average system latency is {average_latency:.2f} ms.",
        f"The platform with the highest average latency is {highest_latency_platform}.",
        f"The platform with the highest error rate is {highest_error_platform}.",
        f"The region with the highest estimated revenue impact is {highest_revenue_impact_region}.",
        f"The highest-priority performance area is {highest_priority_region} - {highest_priority_platform}.",
        f"The highest latency category is {highest_latency_band}.",
        f"The estimated revenue impact associated with failed requests is {estimated_revenue_impact:.2f}.",
        f"The overall profit margin is {profit_margin:.2f}%."
    ]
})


# Step 22: Final Performance Summary

performance_summary = pd.DataFrame({
    "Metric": [
        "Average Latency (ms)",
        "Average Error Rate (%)",
        "Average Conversion Rate (%)",
        "Average Abandonment Rate (%)",
        "Average Quality Score",
        "Average Recovery Time (min)",
        "Total Revenue",
        "Total Profit",
        "Estimated Revenue Impact"
    ],
    "Value": [
        average_latency,
        error_rate,
        conversion_rate,
        abandonment_rate,
        average_quality,
        average_recovery,
        total_revenue,
        total_profit,
        estimated_revenue_impact
    ]
})


# Step 23: Export One Final Analysis File

output_file = "Python Analysis Result.xlsx"

with pd.ExcelWriter(
    output_file,
    engine="openpyxl"
) as writer:

    data_overview.to_excel(
        writer,
        sheet_name="Data Overview",
        index=False
    )

    overall_kpis.to_excel(
        writer,
        sheet_name="Overall KPIs",
        index=False
    )

    monthly_analysis.to_excel(
        writer,
        sheet_name="Monthly Analysis",
        index=False
    )

    region_analysis.to_excel(
        writer,
        sheet_name="Region Analysis",
        index=False
    )

    platform_analysis.to_excel(
        writer,
        sheet_name="Platform Analysis",
        index=False
    )

    version_analysis.to_excel(
        writer,
        sheet_name="Version Analysis",
        index=False
    )

    latency_analysis.to_excel(
        writer,
        sheet_name="Latency Analysis",
        index=False
    )

    latency_bands.to_excel(
        writer,
        sheet_name="Latency Bands",
        index=False
    )

    error_analysis.to_excel(
        writer,
        sheet_name="Error Analysis",
        index=False
    )

    revenue_impact.to_excel(
        writer,
        sheet_name="Revenue Impact",
        index=False
    )

    bottleneck_analysis.to_excel(
        writer,
        sheet_name="Bottleneck Analysis",
        index=False
    )

    quality_analysis.to_excel(
        writer,
        sheet_name="Quality Analysis",
        index=False
    )

    recovery_analysis.to_excel(
        writer,
        sheet_name="Recovery Analysis",
        index=False
    )

    launch_status_analysis.to_excel(
        writer,
        sheet_name="Launch Status",
        index=False
    )

    unit_economics.to_excel(
        writer,
        sheet_name="Unit Economics",
        index=False
    )

    correlation_analysis.to_excel(
        writer,
        sheet_name="Correlation",
        index=True
    )

    high_risk_areas.to_excel(
        writer,
        sheet_name="High Risk Areas",
        index=False
    )

    top_revenue_impact.to_excel(
        writer,
        sheet_name="Top Revenue Impact",
        index=False
    )

    performance_summary.to_excel(
        writer,
        sheet_name="Performance Summary",
        index=False
    )

    insights.to_excel(
        writer,
        sheet_name="Business Insights",
        index=False
    )


# Final Output

print("Analysis Completed Successfully")
print("--------------------------------")

print("Total Users:", total_users)
print("Total Requests:", total_requests)
print("Successful Requests:", successful_requests)
print("Failed Requests:", failed_requests)

print("Conversion Rate:", round(conversion_rate, 2), "%")
print("Abandonment Rate:", round(abandonment_rate, 2), "%")
print("Error Rate:", round(error_rate, 2), "%")

print("Total Revenue:", round(total_revenue, 2))
print("Total Cost:", round(total_cost, 2))
print("Total Profit:", round(total_profit, 2))

print("Average Latency:", round(average_latency, 2), "ms")
print("Average Quality Score:", round(average_quality, 2))
print("Average Recovery Time:", round(average_recovery, 2), "min")

print(
    "Estimated Revenue Impact:",
    round(estimated_revenue_impact, 2)
)

print("\nFinal output file:")
print(output_file)
```
